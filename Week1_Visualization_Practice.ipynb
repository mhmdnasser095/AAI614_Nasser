This notebook demonstrates basic data visualization using Python.


```python
import pandas as pd
import matplotlib.pyplot as plt

# Example community data (illustrative values)
data = {"Community": ["Area A", "Area B", "Area C", "Area D"],
        "Displaced Households": [120, 90, 160, 75]}

df = pd.DataFrame(data)
display(df)

df.plot(kind="bar", x="Community", y="Displaced Households", legend=False)
plt.title("Displaced Households by Community")
plt.ylabel("Number of Households")
plt.show()
```
